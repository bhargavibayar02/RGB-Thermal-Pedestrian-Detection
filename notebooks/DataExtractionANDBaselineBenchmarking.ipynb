```python
import json
import os
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image

base = '/content/drive/MyDrive/hackathon_project/dataset/VTUAV_subset'

# Load annotations
with open(os.path.join(base, 'annotations', 'train.json'), 'r') as f:
    coco = json.load(f)

# Build a lookup: image_id -> list of bboxes
img_id_to_boxes = {}
for ann in coco['annotations']:
    img_id_to_boxes.setdefault(ann['image_id'], []).append(ann['bbox'])

# Build a lookup: file_name -> image_id
filename_to_id = {img['file_name']: img['id'] for img in coco['images']}

def show_pair(file_name):
    img_id = filename_to_id[file_name]
    boxes = img_id_to_boxes.get(img_id, [])

    rgb_path = os.path.join(base, 'VTUAV_co', 'train', 'images', file_name)
    ir_path = os.path.join(base, 'VTUAV_ir', 'train', 'images', file_name)

    rgb_img = Image.open(rgb_path)
    ir_img = Image.open(ir_path)

    fig, axes = plt.subplots(1, 2, figsize=(16, 8))
    for ax, img, title in [(axes[0], rgb_img, 'RGB'), (axes[1], ir_img, 'Thermal')]:
        ax.imshow(img)
        ax.set_title(f'{title} - {file_name} ({len(boxes)} pedestrians)')
        for (x, y, w, h) in boxes:
            rect = patches.Rectangle((x, y), w, h, linewidth=2, edgecolor='lime', facecolor='none')
            ax.add_patch(rect)
        ax.axis('off')
    plt.tight_layout()
    plt.show()

show_pair('00007.jpg')
```

```python
import json
import os
import pandas as pd

base = '/content/drive/MyDrive/hackathon_project/dataset/VTUAV_subset'

def classify_scale(area):
    if area < 32**2:
        return 'small'
    elif area < 96**2:
        return 'medium'
    else:
        return 'large'

summary_rows = []
all_stats = {}

for split in ['train', 'val', 'test']:
    with open(os.path.join(base, 'annotations', f'{split}.json'), 'r') as f:
        coco = json.load(f)

    num_images = len(coco['images'])
    num_annotations = len(coco['annotations'])
    avg_per_image = num_annotations / num_images if num_images > 0 else 0

    scale_counts = {'small': 0, 'medium': 0, 'large': 0}
    for ann in coco['annotations']:
        scale_counts[classify_scale(ann['area'])] += 1

    # Image resolution check
    resolutions = set((img['width'], img['height']) for img in coco['images'])

    all_stats[split] = {
        'num_images': num_images,
        'num_annotations': num_annotations,
        'avg_per_image': avg_per_image,
        'scale_counts': scale_counts,
        'resolutions': resolutions,
    }

    summary_rows.append({
        'Split': split,
        'Images': num_images,
        'Total Pedestrians': num_annotations,
        'Avg per Image': round(avg_per_image, 2),
        'Small': scale_counts['small'],
        'Medium': scale_counts['medium'],
        'Large': scale_counts['large'],
        'Small %': round(100 * scale_counts['small'] / num_annotations, 1),
        'Medium %': round(100 * scale_counts['medium'] / num_annotations, 1),
        'Large %': round(100 * scale_counts['large'] / num_annotations, 1),
    })

    print(f'--- {split.upper()} ---')
    print('Image resolutions found:', resolutions)
    print()

df = pd.DataFrame(summary_rows)
print(df.to_string(index=False))

# Save as CSV for the report later
save_path = '/content/drive/MyDrive/hackathon_project/results/dataset_statistics.csv'
df.to_csv(save_path, index=False)
print('\nSaved statistics table to:', save_path)
```

```python
import numpy as np
from PIL import Image
import json
import os

base = '/content/drive/MyDrive/hackathon_project/dataset/VTUAV_subset'

with open(os.path.join(base, 'annotations', 'train.json'), 'r') as f:
    coco = json.load(f)

# Map image_id -> file_name and image_id -> list of annotations
id_to_filename = {img['id']: img['file_name'] for img in coco['images']}
img_id_to_anns = {}
for ann in coco['annotations']:
    img_id_to_anns.setdefault(ann['image_id'], []).append(ann)

# 1. Find images with the smallest single pedestrian (tiniest)
min_area_per_image = []
for img_id, anns in img_id_to_anns.items():
    smallest = min(a['area'] for a in anns)
    min_area_per_image.append((smallest, img_id, id_to_filename[img_id]))
min_area_per_image.sort()

print('--- Top 5 images with the TINIEST pedestrians ---')
for area, img_id, fname in min_area_per_image[:5]:
    print(f'  {fname}: smallest pedestrian area = {area} px² (side ≈ {area**0.5:.1f}px)')

# 2. Find most crowded images
crowd_counts = [(len(anns), img_id, id_to_filename[img_id]) for img_id, anns in img_id_to_anns.items()]
crowd_counts.sort(reverse=True)

print('\n--- Top 5 MOST CROWDED images ---')
for count, img_id, fname in crowd_counts[:5]:
    print(f'  {fname}: {count} pedestrians')

# 3. Find darkest (lowest average brightness) RGB images -- sample 200 for speed
import random
random.seed(42)
sample_ids = random.sample(list(img_id_to_anns.keys()), min(200, len(img_id_to_anns)))

brightness_list = []
for img_id in sample_ids:
    fname = id_to_filename[img_id]
    path = os.path.join(base, 'VTUAV_co', 'train', 'images', fname)
    img = Image.open(path).convert('L')  # grayscale for brightness
    avg_brightness = np.array(img).mean()
    brightness_list.append((avg_brightness, img_id, fname))
brightness_list.sort()

print('\n--- Top 5 DARKEST images (from 200-image sample) ---')
for brightness, img_id, fname in brightness_list[:5]:
    print(f'  {fname}: avg brightness = {brightness:.1f} / 255')

print('\n--- Top 5 BRIGHTEST images (for contrast) ---')
for brightness, img_id, fname in brightness_list[-5:]:
    print(f'  {fname}: avg brightness = {brightness:.1f} / 255')
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os

base = '/content/drive/MyDrive/hackathon_project/dataset/VTUAV_subset'

def apply_clahe(img_array, clip_limit=2.0, grid_size=(8,8)):
    if len(img_array.shape) == 3:  # color image
        lab = cv2.cvtColor(img_array, cv2.COLOR_RGB2LAB)
        l, a, b = cv2.split(lab)
        clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=grid_size)
        l_enhanced = clahe.apply(l)
        enhanced = cv2.merge([l_enhanced, a, b])
        return cv2.cvtColor(enhanced, cv2.COLOR_LAB2RGB)
    else:  # grayscale (thermal)
        clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=grid_size)
        return clahe.apply(img_array)

test_file = '08848.jpg'  # our darkest RGB image found earlier
rgb_path = os.path.join(base, 'VTUAV_co', 'train', 'images', test_file)
ir_path = os.path.join(base, 'VTUAV_ir', 'train', 'images', test_file)

rgb_arr = np.array(Image.open(rgb_path).convert('RGB'))
ir_arr = np.array(Image.open(ir_path).convert('L'))

rgb_enhanced = apply_clahe(rgb_arr)
ir_enhanced = apply_clahe(ir_arr)

fig, axes = plt.subplots(2, 2, figsize=(14, 12))
axes[0,0].imshow(rgb_arr); axes[0,0].set_title('RGB - Original'); axes[0,0].axis('off')
axes[0,1].imshow(rgb_enhanced); axes[0,1].set_title('RGB - CLAHE Enhanced'); axes[0,1].axis('off')
axes[1,0].imshow(ir_arr, cmap='gray'); axes[1,0].set_title('Thermal - Original'); axes[1,0].axis('off')
axes[1,1].imshow(ir_enhanced, cmap='gray'); axes[1,1].set_title('Thermal - CLAHE Enhanced'); axes[1,1].axis('off')
plt.tight_layout()

save_path = '/content/drive/MyDrive/hackathon_project/results/preprocessing_demo.png'
plt.savefig(save_path, dpi=120, bbox_inches='tight')
print('Saved:', save_path)
plt.show()
```

```python
loading_path = '/content/drive/MyDrive/hackathon_project/baseline/mmdet-rgbtdroneperson/mmdet/datasets/pipelines/loading.py'

new_class_code = '''

@PIPELINES.register_module()
class LoadImagePairFromFileBlank(LoadImagePairFromFile):
    """Load a dualspectral image pair from file, then zero out ("blank")
    one of the two modalities. Used for unimodal ablation testing of a
    fusion model without retraining from scratch: the model still receives
    two tensors of the correct shape, but one carries no real signal
    (all-zero), so the fusion module effectively only "sees" the other.

    Args:
        blank_modality (str): Which modality to zero out.
            'first' zeros spectrals[0] (e.g. RGB, for a Thermal-only test).
            'second' zeros spectrals[1] (e.g. thermal, for an RGB-only test).
        Other args are identical to LoadImagePairFromFile.
    """

    def __init__(self, blank_modality, **kwargs):
        assert blank_modality in ('first', 'second'), \\
            "blank_modality must be 'first' or 'second'"
        self.blank_modality = blank_modality
        super().__init__(**kwargs)

    def __call__(self, results):
        results = super().__call__(results)
        img1, img2 = results['img']
        if self.blank_modality == 'first':
            img1 = np.zeros_like(img1)
        else:
            img2 = np.zeros_like(img2)
        results['img'] = (img1, img2)
        return results
'''

with open(loading_path, 'r') as f:
    content = f.read()

assert 'LoadImagePairFromFileBlank' not in content, "Already added — skip this step"

with open(loading_path, 'a') as f:
    f.write(new_class_code)

print("Class appended successfully.")
```

```python
import torch
import torch.nn as nn

class ChannelAttention(nn.Module):

    def __init__(self, channels, reduction=16):
        super().__init__()

        self.avg_pool = nn.AdaptiveAvgPool2d(1)

        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels),
            nn.Sigmoid()
        )

    def forward(self, x):

        b, c, h, w = x.size()

        y = self.avg_pool(x)

        y = y.view(b, c)

        y = self.fc(y)

        y = y.view(b, c, 1, 1)

        return x * y
```

```python
# Phase 3.1 - Channel Attention Module
import torch
import torch.nn as nn
class ChannelAttention(nn.Module):
    """
    Channel Attention Module
    Learns which feature channels are important.
    """

    def __init__(self, channels, reduction=16):
        super(ChannelAttention, self).__init__()

        # Global Average Pooling
        self.avg_pool = nn.AdaptiveAvgPool2d(1)

        # Fully Connected Layers
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels),
            nn.Sigmoid()
        )

    def forward(self, x):

        # Input shape
        batch_size, channels, height, width = x.size()

        # Step 1: Global Average Pooling
        y = self.avg_pool(x)

        # Shape: (B, C)
        y = y.view(batch_size, channels)

        # Step 2: Learn channel importance
        y = self.fc(y)

        # Shape: (B, C, 1, 1)
        y = y.view(batch_size, channels, 1, 1)

        # Step 3: Apply attention
        out = x * y

        return out
```

```python
class SpatialAttention(nn.Module):
    """
    Spatial Attention Module
    Learns where to focus in the feature map.
    """

    def __init__(self, kernel_size=7):
        super(SpatialAttention, self).__init__()

        self.conv = nn.Conv2d(
            in_channels=2,
            out_channels=1,
            kernel_size=kernel_size,
            padding=kernel_size // 2,
            bias=False
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        # Average across channels
        avg_out = torch.mean(x, dim=1, keepdim=True)

        # Maximum across channels
        max_out, _ = torch.max(x, dim=1, keepdim=True)

        # Combine
        y = torch.cat([avg_out, max_out], dim=1)

        # Learn attention
        y = self.conv(y)

        y = self.sigmoid(y)

        # Apply attention
        out = x * y

        return out
```

```python
class CrossModalAttention(nn.Module):
    """
    Cross Modal Attention
    Allows RGB and Thermal features
    to guide each other.
    """

    def __init__(self, channels):
        super(CrossModalAttention, self).__init__()

        self.attention = nn.Sequential(

            nn.Conv2d(
                channels,
                channels,
                kernel_size=3,
                padding=1,
                bias=False
            ),

            nn.BatchNorm2d(channels),

            nn.ReLU(inplace=True),

            nn.Conv2d(
                channels,
                channels,
                kernel_size=1
            ),

            nn.Sigmoid()

        )

    def forward(self, rgb, thermal):

        # Shared information
        shared = rgb + thermal

        # Learn attention
        attention = self.attention(shared)

        # Enhance both modalities
        rgb_out = rgb * attention

        thermal_out = thermal * attention

        return rgb_out, thermal_out
```

```python
%%writefile /content/drive/MyDrive/hackathon_project/baseline/mmdet-rgbtdroneperson/mmdet/models/detectors/qfdet.py

# Copyright (c) OpenMMLab. All rights reserved.
from ..builder import DETECTORS, build_backbone, build_head
from .single_stage import SingleStageDetector
import torch
import torch.nn as nn
import torch.nn.functional as F
from mmdet.core import bbox2result, multi_apply


@DETECTORS.register_module()
class QFDet(SingleStageDetector):
    """Implementation of `QFDet`."""

    def __init__(self,
                 backbone,
                 neck,
                 bbox_head,
                 bbox_prehead,
                 base_fusion='cat',
                 quality_attention=True,
                 poolupsample=None,
                 reweight=False,
                 train_cfg=None,
                 test_cfg=None,
                 pretrained=None,
                 init_cfg=None):
        super(QFDet, self).__init__(backbone, neck, bbox_head, train_cfg,
                                   test_cfg, pretrained, init_cfg)
        self.backbone_t = build_backbone(backbone)
        self.base_fusion = base_fusion
        self.fuse = Fusion_strategy(neck['out_channels'])
        bbox_prehead.update(train_cfg=train_cfg)
        bbox_prehead.update(test_cfg=test_cfg)
        self.bbox_prehead = build_head(bbox_prehead)
        self.iter = 0
        if poolupsample is not None:
            self.poolupsample = PoolingUpsample(neck['out_channels'])
        else:
            self.poolupsample = None
        self.quality_attention = quality_attention

        self.reweight = reweight

    def extract_feat(self, img):
        """Directly extract features from the backbone+neck."""
        # self.iter = self.iter + 1
        v_img, t_img = img
        v_feats = self.backbone(v_img)
        t_feats = self.backbone_t(t_img)
        if self.with_neck:
            v_feats = self.neck(v_feats)
            t_feats = self.neck(t_feats)

        return (v_feats, t_feats)

    def forward_dummy(self, img):
        """Used for computing network flops.

        See `mmdetection/tools/analysis_tools/get_flops.py`
        """
        x = self.extract_feat(img)
        x_vs, x_ts = x
        num_level = len(x_vs)
        quality_preds_t, quality_preds_v = self.bbox_prehead.forward_test(x)
        fused_x = []
        for i in range(num_level):
            quality_pred_t, quality_pred_v = my_norm(quality_preds_t[i], quality_preds_v[i], type='minmax')

            x_t = (1 + quality_pred_t) * x_ts[i]
            x_v = (1 + quality_pred_v) * x_vs[i]

            if self.poolupsample is not None and i < num_level-1:
                x_v = self.poolupsample(x_v)

            fused_x_ = self.fuse(x_t, x_v, 'cat')
            fused_x.append(fused_x_)
        outs = self.bbox_head(fused_x)
        return outs

    def forward_train(self,
                      img,
                      img_metas,
                      gt_bboxes,
                      gt_labels,
                      gt_bboxes_ignore=None):
        """
        Args:
            img (Tensor): Input images of shape (N, C, H, W).
                Typically these should be mean centered and std scaled.
            img_metas (list[dict]): A List of image info dict where each dict
                has: 'img_shape', 'scale_factor', 'flip', and may also contain
                'filename', 'ori_shape', 'pad_shape', and 'img_norm_cfg'.
                For details on the values of these keys see
                :class:`mmdet.datasets.pipelines.Collect`.
            gt_bboxes (list[Tensor]): Each item are the truth boxes for each
                image in [tl_x, tl_y, br_x, br_y] format.
            gt_labels (list[Tensor]): Class indices corresponding to each box
            gt_bboxes_ignore (None | list[Tensor]): Specify which bounding
                boxes can be ignored when computing the loss.

        Returns:
            dict[str, Tensor]: A dictionary of loss components.
        """
        # super(SingleStageDetector, self).forward_train(img, img_metas)
        losses = dict()
        x = self.extract_feat(img)
        x_vs, x_ts = x

        cls_scores_t, bbox_preds_t, centernesses_t, quality_preds_t, \
        cls_scores_v, bbox_preds_v, centernesses_v, quality_preds_v = self.bbox_prehead(x)

        pre_loss = self.bbox_prehead.loss(
             cls_scores_t,
             bbox_preds_t,
             centernesses_t,
             quality_preds_t,
             cls_scores_v,
             bbox_preds_v,
             centernesses_v,
             quality_preds_v,
             gt_bboxes,
             gt_labels,
             img_metas,
             gt_bboxes_ignore=None)
        losses.update(pre_loss)

        fused_x = self.qce_fusion(x, quality_preds_t, quality_preds_t)

        loss = self.bbox_head.forward_train(fused_x, img_metas, gt_bboxes,
                                              gt_labels, gt_bboxes_ignore)
        if self.reweight:
            weight = torch.linspace(1, 5, steps=len(x_ts), device=loss['loss_cls'][0].device)
            for k in loss.keys():
                for i in range(weight.size()[0]):
                    loss[k][i] = weight[i]*loss[k][i]
        losses.update(loss)
        return losses

    def simple_test(self, img, img_metas, rescale=False):
        """Test function without test-time augmentation.

        Args:
            img (torch.Tensor): Images with shape (N, C, H, W).
            img_metas (list[dict]): List of image information.
            rescale (bool, optional): Whether to rescale the results.
                Defaults to False.

        Returns:
            list[list[np.ndarray]]: BBox results of each image and classes.
                The outer list corresponds to each image. The inner list
                corresponds to each class.
        """
        self.iter += 1
        x = self.extract_feat(img)

        quality_preds_t, quality_preds_v = self.bbox_prehead.forward_test(x)

        fused_x = self.qce_fusion(x, quality_preds_t, quality_preds_v)

        results_list = self.bbox_head.simple_test(
            fused_x, img_metas, rescale=rescale)

        bbox_results = [
            bbox2result(det_bboxes, det_labels, self.bbox_head.num_classes)
            for det_bboxes, det_labels in results_list
        ]
        return bbox_results

    def qce_fusion(self, x, quality_t, quality_v):
        x_vs, x_ts = x

        num_level = len(x_vs)

        fused_x = []
        for i in range(num_level):
            x_t = x_ts[i]
            x_v = x_vs[i]
            if self.quality_attention:
                quality_pred_t = torch.max(quality_t[i], dim=1, keepdim=True)[0]
                quality_pred_v = torch.max(quality_v[i], dim=1, keepdim=True)[0]

                quality_pred_t, quality_pred_v = my_norm(quality_pred_t, quality_pred_v, type='minmax')

                x_t = (1 + quality_pred_t) * x_t
                x_v = (1 + quality_pred_v) * x_v

            if self.poolupsample is not None and i < num_level-1:
                x_v = self.poolupsample(x_v)
                # x_t = self.poolupsample(x_t)

            fused_x_ = self.fuse(x_t, x_v, self.base_fusion)

            fused_x.append(fused_x_)

        return fused_x


def save_feature_to_img(features, name, timestamp, method='cv2', channel=None, output_dir=None, maxmin=None):
    import numpy as np
    import cv2
    import matplotlib.pyplot as plt
    import os

    if output_dir is None:
        output_dir = '/home/zhangyan22/mmdetection/work_dirs/atss_r50_fpn_hf_qprelchead_1x_dp/vis_wo_optimize_v'
    if not os.path.exists(output_dir):
        os.mkdir(output_dir)
    if not isinstance(timestamp, str):
        timestamp = str(timestamp)

    # for i in range(len(features)):
    if isinstance(features, list) or isinstance(features, tuple):
        for i in range(3):
            features_ = features[i]
            for j in range(features_.shape[0]):
                upsample = nn.Upsample(scale_factor=4, mode='bilinear', align_corners=True)
                features_ = upsample(features_)

                feature = features_[j, :, :, :]
                if channel is None:
                    feature = torch.sum(feature, 0)
                else:
                    feature = feature[channel, :, :]
                feature = feature.detach().cpu().numpy() # 转为numpy

                dist_dir = os.path.join(output_dir, timestamp)
                if not os.path.exists(dist_dir):
                    os.mkdir(dist_dir)

                if method == 'cv2':
                    if maxmin is not None:
                        img = (feature - maxmin[1])/(maxmin[0] - maxmin[1] + 1e-5) * 255
                    else:
                        img = (feature - np.amin(feature))/(np.amax(feature) - np.amin(feature) + 1e-5) * 255 # 注意要防止分母为0！
                    img = img.astype(np.uint8)
                    img = cv2.applyColorMap(img, cv2.COLORMAP_JET)
                    plt.imshow(feature)
                    plt.axis('off')
                    cv2.imwrite(os.path.join(dist_dir, name + str(i) + '.jpg'), img)

                elif method == 'matshow':
                    plt.matshow(feature, interpolation='nearest')
                    plt.colorbar()
                    plt.axis('off')

                    plt.savefig(os.path.join(dist_dir, name + str(i) + '.png'))
                    plt.close()
                else:
                    NotImplementedError()

    else:
        for j in range(features.shape[0]):
            upsample = nn.Upsample(scale_factor=4, mode='bilinear', align_corners=True)
            features = upsample(features)

            feature = features[j, :, :, :]
            if channel is None:
                feature = torch.sum(feature, 0)
            else:
                feature = feature[channel, :, :]
            feature = feature.detach().cpu().numpy() # 转为numpy

            dist_dir = os.path.join(output_dir, timestamp)
            if not os.path.exists(dist_dir):
                os.mkdir(dist_dir)

            if method == 'cv2':
                if maxmin is not None:
                    img = (feature - maxmin[1])/(maxmin[0] - maxmin[1] + 1e-5) * 255
                else:
                    img = (feature - np.amin(feature))/(np.amax(feature) - np.amin(feature) + 1e-5) * 255 # 注意要防止分母为0！
                img = img.astype(np.uint8)
                img = cv2.applyColorMap(img, cv2.COLORMAP_JET)
                plt.imshow(feature)
                plt.axis('off')
                cv2.imwrite(os.path.join(dist_dir, name + '.jpg'), img)

            elif method == 'matshow':
                plt.matshow(feature, interpolation='nearest')
                plt.colorbar()
                plt.axis('off')

                plt.savefig(os.path.join(dist_dir, name + '.png'))
                plt.close()
            else:
                NotImplementedError()


def feature_map_norm(x):
    # input (B, C, H, W)
    bs, c, h, w = x.shape
    x = x.view(bs, -1, h*w)
    x_mean = torch.mean(x, dim=2, keepdim=True)
    x_std = torch.std(x, dim=2, keepdim=True)
    x = (x - x_mean) / x_std
    x = x.view(bs, c, h, w)
    return x

def my_norm(x1, x2, type='standard'):
    assert type in ['standard', 'minmax']
    bs, _ , H, W = x1.size()
    _, _, h, w = x2.size()
    x1 = x1.view(bs, -1, H*W)
    x2 = x2.view(bs, -1, h*w)
    concat = torch.cat((x1, x2), dim=2)
    if type == 'standard':
        x_mean = torch.mean(concat, dim=2, keepdim=True)
        x_std = torch.std(concat, dim=2, keepdim=True)
        x1 = (x1 - x_mean) / x_std
        x2 = (x2 - x_mean) / x_std
    elif type == 'minmax':
        x_min = torch.min(concat, dim=2, keepdim=True)[0]
        x_max = torch.max(concat, dim=2, keepdim=True)[0]
        x1 = (x1 - x_min) / x_max
        x2 = (x2 - x_min) / x_max
    x1 = x1.view(bs, -1, H, W)
    x2 = x2.view(bs, -1, h, w)
    return [x1, x2]


class Fusion_ADD(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = en_ir + en_vi
        return temp

class Fusion_AVG(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = (en_ir + en_vi) / 2
        return temp

class Fusion_MAX(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = torch.max(en_ir, en_vi)
        return temp

class Fusion_CAT(torch.nn.Module):
    def __init__(self, in_channels) -> None:
        super().__init__()
        self.conv1x1 = nn.Conv2d(2*in_channels, in_channels, 1)

    def forward(self, en_ir, en_vi):
        temp = torch.cat((en_ir, en_vi), 1)
        temp = self.conv1x1(temp)
        return temp


class Fusion_FLF(nn.Module):
    """Feature-Level Fusion: channel-attention gated fusion.

    Concatenates v/t features, computes a per-channel gate via
    squeeze-excitation on the concatenated tensor, reweights each
    modality's channels, then projects back to in_channels.
    """
    def __init__(self, in_channels, reduction=16):
        super().__init__()
        self.in_channels = in_channels
        hidden = max(in_channels // reduction, 8)
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Conv2d(in_channels * 2, hidden, 1, bias=False),
            nn.ReLU(inplace=True),
            nn.Conv2d(hidden, in_channels * 2, 1, bias=False),
        )
        self.sigmoid = nn.Sigmoid()
        self.conv_out = nn.Conv2d(in_channels * 2, in_channels, 1)

    def forward(self, en_ir, en_vi):
        concat = torch.cat((en_ir, en_vi), dim=1)      # (B, 2C, H, W)
        pooled = self.avg_pool(concat)                  # (B, 2C, 1, 1)
        gate = self.sigmoid(self.fc(pooled))             # (B, 2C, 1, 1)
        weighted = concat * gate                         # (B, 2C, H, W)
        fused = self.conv_out(weighted)                  # (B, C, H, W)
        return fused


# Fusion strategy, two type
class Fusion_strategy(nn.Module):
    def __init__(self, in_channels):
        super(Fusion_strategy, self).__init__()
        self.fusion_add = Fusion_ADD()
        self.fusion_avg = Fusion_AVG()
        self.fusion_max = Fusion_MAX()
        self.fusion_cat = Fusion_CAT(in_channels=in_channels)
        self.fusion_flf = Fusion_FLF(in_channels=in_channels)

    def forward(self, v_feat, t_feat, fs_type):
        self.fs_type = fs_type
        if self.fs_type == 'add':
            fusion_operation = self.fusion_add
        elif self.fs_type == 'avg':
            fusion_operation = self.fusion_avg
        elif self.fs_type == 'max':
            fusion_operation = self.fusion_max
        elif self.fs_type == 'cat':
            fusion_operation = self.fusion_cat
        elif self.fs_type == 'flf':
            fusion_operation = self.fusion_flf
        else:
            raise ValueError(f"Unknown fs_type: {fs_type}")

        if isinstance(v_feat, tuple) or isinstance(v_feat, list):
            fused_feat = []
            for i in range(len(v_feat)):
                fused_feat.append(fusion_operation(v_feat[i], t_feat[i]))
        else:
            fused_feat = fusion_operation(v_feat, t_feat)

        return fused_feat



class PoolingUpsample(nn.Module):
    def __init__(self, in_channels):
        super().__init__()
        self.in_channels = in_channels
        self.maxpooling = nn.MaxPool2d(2, 2, dilation=1)
        # self.upsample = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        self.conv1x1 = nn.Conv2d(in_channels*2, in_channels, 1)

    def forward(self, x):
        x_ = self.maxpooling(x)
        # x_ = self.upsample(x_)
        x_ = F.interpolate(x_, mode='bilinear', size=x.shape[-2:], align_corners=True)
        # import pdb; pdb.set_trace()
        x = self.conv1x1(torch.cat((x, x_), 1))
        return x
```

```python
%%writefile /content/drive/MyDrive/hackathon_project/baseline/mmdet-rgbtdroneperson/mmdet/models/detectors/qfdet.py
# Copyright (c) OpenMMLab. All rights reserved.
from ..builder import DETECTORS, build_backbone, build_head
from .single_stage import SingleStageDetector
import torch
import torch.nn as nn
import torch.nn.functional as F
from mmdet.core import bbox2result, multi_apply


@DETECTORS.register_module()
class QFDet(SingleStageDetector):
    """Implementation of `QFDet`."""

    def __init__(self,
                 backbone,
                 neck,
                 bbox_head,
                 bbox_prehead,
                 base_fusion='cat',
                 quality_attention=True,
                 poolupsample=None,
                 reweight=False,
                 train_cfg=None,
                 test_cfg=None,
                 pretrained=None,
                 init_cfg=None):
        super(QFDet, self).__init__(backbone, neck, bbox_head, train_cfg,
                                   test_cfg, pretrained, init_cfg)
        self.backbone_t = build_backbone(backbone)
        self.base_fusion = base_fusion
        self.fuse = Fusion_strategy(neck['out_channels'])
        bbox_prehead.update(train_cfg=train_cfg)
        bbox_prehead.update(test_cfg=test_cfg)
        self.bbox_prehead = build_head(bbox_prehead)
        self.iter = 0
        if poolupsample is not None:
            self.poolupsample = PoolingUpsample(neck['out_channels'])
        else:
            self.poolupsample = None
        self.quality_attention = quality_attention

        self.reweight = reweight

    def extract_feat(self, img):
        """Directly extract features from the backbone+neck."""
        # self.iter = self.iter + 1
        v_img, t_img = img
        v_feats = self.backbone(v_img)
        t_feats = self.backbone_t(t_img)
        if self.with_neck:
            v_feats = self.neck(v_feats)
            t_feats = self.neck(t_feats)

        return (v_feats, t_feats)

    def forward_dummy(self, img):
        """Used for computing network flops.

        See `mmdetection/tools/analysis_tools/get_flops.py`
        """
        x = self.extract_feat(img)
        x_vs, x_ts = x
        num_level = len(x_vs)
        quality_preds_t, quality_preds_v = self.bbox_prehead.forward_test(x)
        fused_x = []
        for i in range(num_level):
            quality_pred_t, quality_pred_v = my_norm(quality_preds_t[i], quality_preds_v[i], type='minmax')

            x_t = (1 + quality_pred_t) * x_ts[i]
            x_v = (1 + quality_pred_v) * x_vs[i]

            if self.poolupsample is not None and i < num_level-1:
                x_v = self.poolupsample(x_v)

            fused_x_ = self.fuse(x_t, x_v, 'cat')
            fused_x.append(fused_x_)
        outs = self.bbox_head(fused_x)
        return outs

    def forward_train(self,
                      img,
                      img_metas,
                      gt_bboxes,
                      gt_labels,
                      gt_bboxes_ignore=None):
        """
        Args:
            img (Tensor): Input images of shape (N, C, H, W).
                Typically these should be mean centered and std scaled.
            img_metas (list[dict]): A List of image info dict where each dict
                has: 'img_shape', 'scale_factor', 'flip', and may also contain
                'filename', 'ori_shape', 'pad_shape', and 'img_norm_cfg'.
                For details on the values of these keys see
                :class:`mmdet.datasets.pipelines.Collect`.
            gt_bboxes (list[Tensor]): Each item are the truth boxes for each
                image in [tl_x, tl_y, br_x, br_y] format.
            gt_labels (list[Tensor]): Class indices corresponding to each box
            gt_bboxes_ignore (None | list[Tensor]): Specify which bounding
                boxes can be ignored when computing the loss.

        Returns:
            dict[str, Tensor]: A dictionary of loss components.
        """
        # super(SingleStageDetector, self).forward_train(img, img_metas)
        losses = dict()
        x = self.extract_feat(img)
        x_vs, x_ts = x

        cls_scores_t, bbox_preds_t, centernesses_t, quality_preds_t, \
        cls_scores_v, bbox_preds_v, centernesses_v, quality_preds_v = self.bbox_prehead(x)

        pre_loss = self.bbox_prehead.loss(
             cls_scores_t,
             bbox_preds_t,
             centernesses_t,
             quality_preds_t,
             cls_scores_v,
             bbox_preds_v,
             centernesses_v,
             quality_preds_v,
             gt_bboxes,
             gt_labels,
             img_metas,
             gt_bboxes_ignore=None)
        losses.update(pre_loss)

        fused_x = self.qce_fusion(x, quality_preds_t, quality_preds_t)

        loss = self.bbox_head.forward_train(fused_x, img_metas, gt_bboxes,
                                              gt_labels, gt_bboxes_ignore)
        if self.reweight:
            weight = torch.linspace(1, 5, steps=len(x_ts), device=loss['loss_cls'][0].device)
            for k in loss.keys():
                for i in range(weight.size()[0]):
                    loss[k][i] = weight[i]*loss[k][i]
        losses.update(loss)
        return losses

    def simple_test(self, img, img_metas, rescale=False):
        """Test function without test-time augmentation.

        Args:
            img (torch.Tensor): Images with shape (N, C, H, W).
            img_metas (list[dict]): List of image information.
            rescale (bool, optional): Whether to rescale the results.
                Defaults to False.

        Returns:
            list[list[np.ndarray]]: BBox results of each image and classes.
                The outer list corresponds to each image. The inner list
                corresponds to each class.
        """
        self.iter += 1
        x = self.extract_feat(img)

        quality_preds_t, quality_preds_v = self.bbox_prehead.forward_test(x)

        fused_x = self.qce_fusion(x, quality_preds_t, quality_preds_v)

        results_list = self.bbox_head.simple_test(
            fused_x, img_metas, rescale=rescale)

        bbox_results = [
            bbox2result(det_bboxes, det_labels, self.bbox_head.num_classes)
            for det_bboxes, det_labels in results_list
        ]
        return bbox_results

    def qce_fusion(self, x, quality_t, quality_v):
        x_vs, x_ts = x

        num_level = len(x_vs)

        fused_x = []
        for i in range(num_level):
            x_t = x_ts[i]
            x_v = x_vs[i]
            if self.quality_attention:
                quality_pred_t = torch.max(quality_t[i], dim=1, keepdim=True)[0]
                quality_pred_v = torch.max(quality_v[i], dim=1, keepdim=True)[0]

                quality_pred_t, quality_pred_v = my_norm(quality_pred_t, quality_pred_v, type='minmax')

                x_t = (1 + quality_pred_t) * x_t
                x_v = (1 + quality_pred_v) * x_v

            if self.poolupsample is not None and i < num_level-1:
                x_v = self.poolupsample(x_v)
                # x_t = self.poolupsample(x_t)

            fused_x_ = self.fuse(x_t, x_v, self.base_fusion)

            fused_x.append(fused_x_)

        return fused_x


def save_feature_to_img(features, name, timestamp, method='cv2', channel=None, output_dir=None, maxmin=None):
    import numpy as np
    import cv2
    import matplotlib.pyplot as plt
    import os

    if output_dir is None:
        output_dir = '/home/zhangyan22/mmdetection/work_dirs/atss_r50_fpn_hf_qprelchead_1x_dp/vis_wo_optimize_v'
    if not os.path.exists(output_dir):
        os.mkdir(output_dir)
    if not isinstance(timestamp, str):
        timestamp = str(timestamp)

    # for i in range(len(features)):
    if isinstance(features, list) or isinstance(features, tuple):
        for i in range(3):
            features_ = features[i]
            for j in range(features_.shape[0]):
                upsample = nn.Upsample(scale_factor=4, mode='bilinear', align_corners=True)
                features_ = upsample(features_)

                feature = features_[j, :, :, :]
                if channel is None:
                    feature = torch.sum(feature, 0)
                else:
                    feature = feature[channel, :, :]
                feature = feature.detach().cpu().numpy() # 转为numpy

                dist_dir = os.path.join(output_dir, timestamp)
                if not os.path.exists(dist_dir):
                    os.mkdir(dist_dir)

                if method == 'cv2':
                    if maxmin is not None:
                        img = (feature - maxmin[1])/(maxmin[0] - maxmin[1] + 1e-5) * 255
                    else:
                        img = (feature - np.amin(feature))/(np.amax(feature) - np.amin(feature) + 1e-5) * 255 # 注意要防止分母为0！
                    img = img.astype(np.uint8)
                    img = cv2.applyColorMap(img, cv2.COLORMAP_JET)
                    plt.imshow(feature)
                    plt.axis('off')
                    cv2.imwrite(os.path.join(dist_dir, name + str(i) + '.jpg'), img)

                elif method == 'matshow':
                    plt.matshow(feature, interpolation='nearest')
                    plt.colorbar()
                    plt.axis('off')

                    plt.savefig(os.path.join(dist_dir, name + str(i) + '.png'))
                    plt.close()
                else:
                    NotImplementedError()

    else:
        for j in range(features.shape[0]):
            upsample = nn.Upsample(scale_factor=4, mode='bilinear', align_corners=True)
            features = upsample(features)

            feature = features[j, :, :, :]
            if channel is None:
                feature = torch.sum(feature, 0)
            else:
                feature = feature[channel, :, :]
            feature = feature.detach().cpu().numpy() # 转为numpy

            dist_dir = os.path.join(output_dir, timestamp)
            if not os.path.exists(dist_dir):
                os.mkdir(dist_dir)

            if method == 'cv2':
                if maxmin is not None:
                    img = (feature - maxmin[1])/(maxmin[0] - maxmin[1] + 1e-5) * 255
                else:
                    img = (feature - np.amin(feature))/(np.amax(feature) - np.amin(feature) + 1e-5) * 255 # 注意要防止分母为0！
                img = img.astype(np.uint8)
                img = cv2.applyColorMap(img, cv2.COLORMAP_JET)
                plt.imshow(feature)
                plt.axis('off')
                cv2.imwrite(os.path.join(dist_dir, name + '.jpg'), img)

            elif method == 'matshow':
                plt.matshow(feature, interpolation='nearest')
                plt.colorbar()
                plt.axis('off')

                plt.savefig(os.path.join(dist_dir, name + '.png'))
                plt.close()
            else:
                NotImplementedError()


def feature_map_norm(x):
    # input (B, C, H, W)
    bs, c, h, w = x.shape
    x = x.view(bs, -1, h*w)
    x_mean = torch.mean(x, dim=2, keepdim=True)
    x_std = torch.std(x, dim=2, keepdim=True)
    x = (x - x_mean) / x_std
    x = x.view(bs, c, h, w)
    return x

def my_norm(x1, x2, type='standard'):
    assert type in ['standard', 'minmax']
    bs, _ , H, W = x1.size()
    _, _, h, w = x2.size()
    x1 = x1.view(bs, -1, H*W)
    x2 = x2.view(bs, -1, h*w)
    concat = torch.cat((x1, x2), dim=2)
    if type == 'standard':
        x_mean = torch.mean(concat, dim=2, keepdim=True)
        x_std = torch.std(concat, dim=2, keepdim=True)
        x1 = (x1 - x_mean) / x_std
        x2 = (x2 - x_mean) / x_std
    elif type == 'minmax':
        x_min = torch.min(concat, dim=2, keepdim=True)[0]
        x_max = torch.max(concat, dim=2, keepdim=True)[0]
        x1 = (x1 - x_min) / x_max
        x2 = (x2 - x_min) / x_max
    x1 = x1.view(bs, -1, H, W)
    x2 = x2.view(bs, -1, h, w)
    return [x1, x2]


class Fusion_ADD(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = en_ir + en_vi
        return temp

class Fusion_AVG(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = (en_ir + en_vi) / 2
        return temp

class Fusion_MAX(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = torch.max(en_ir, en_vi)
        return temp

class Fusion_CAT(torch.nn.Module):
    def __init__(self, in_channels) -> None:
        super().__init__()
        self.conv1x1 = nn.Conv2d(2*in_channels, in_channels, 1)

    def forward(self, en_ir, en_vi):
        temp = torch.cat((en_ir, en_vi), 1)
        temp = self.conv1x1(temp)
        return temp


class Fusion_FLF(nn.Module):
    """Feature-Level Fusion: channel-attention gated fusion.

    Concatenates v/t features, computes a per-channel gate via
    squeeze-excitation on the concatenated tensor, reweights each
    modality's channels, then projects back to in_channels.
    """
    def __init__(self, in_channels, reduction=16):
        super().__init__()
        self.in_channels = in_channels
        hidden = max(in_channels // reduction, 8)
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Conv2d(in_channels * 2, hidden, 1, bias=False),
            nn.ReLU(inplace=True),
            nn.Conv2d(hidden, in_channels * 2, 1, bias=False),
        )
        self.sigmoid = nn.Sigmoid()
        self.conv_out = nn.Conv2d(in_channels * 2, in_channels, 1)

    def forward(self, en_ir, en_vi):
        concat = torch.cat((en_ir, en_vi), dim=1)      # (B, 2C, H, W)
        pooled = self.avg_pool(concat)                  # (B, 2C, 1, 1)
        gate = self.sigmoid(self.fc(pooled))             # (B, 2C, 1, 1)
        weighted = concat * gate                         # (B, 2C, H, W)
        fused = self.conv_out(weighted)                  # (B, C, H, W)
        return fused


class CrossModalAttention(nn.Module):
    """One-directional cross attention: query modality attends over key/value modality.
    Pools K/V when spatial size is large, to keep attention affordable at fine FPN levels."""
    def __init__(self, in_channels, reduction=8, pool_threshold=4096):
        super().__init__()
        self.hidden = max(in_channels // reduction, 8)
        self.query_conv = nn.Conv2d(in_channels, self.hidden, 1)
        self.key_conv = nn.Conv2d(in_channels, self.hidden, 1)
        self.value_conv = nn.Conv2d(in_channels, in_channels, 1)
        self.gamma = nn.Parameter(torch.zeros(1))
        self.scale = self.hidden ** -0.5
        self.pool_threshold = pool_threshold

    def forward(self, query_feat, kv_feat):
        B, C, H, W = query_feat.shape
        Q = self.query_conv(query_feat).view(B, self.hidden, H * W).permute(0, 2, 1)  # (B, HW, hidden)

        kv_h, kv_w = kv_feat.shape[2], kv_feat.shape[3]
        kv_src = kv_feat
        if kv_h * kv_w > self.pool_threshold:
            pool_factor = 2
            kv_src = F.avg_pool2d(kv_feat, pool_factor)
            kv_h, kv_w = kv_src.shape[2], kv_src.shape[3]

        K = self.key_conv(kv_src).view(B, self.hidden, kv_h * kv_w)          # (B, hidden, N)
        V = self.value_conv(kv_src).view(B, C, kv_h * kv_w).permute(0, 2, 1)  # (B, N, C)

        attn = torch.bmm(Q, K) * self.scale   # (B, HW, N)
        attn = F.softmax(attn, dim=-1)

        out = torch.bmm(attn, V)              # (B, HW, C)
        out = out.permute(0, 2, 1).view(B, C, H, W)

        return query_feat + self.gamma * out


class Fusion_CMATTN(nn.Module):
    """Cross-Modal Attention Fusion: RGB and Thermal attend to each other's
    features at every spatial location, before being merged. Unlike Fusion_FLF
    (a single global per-channel gate), this lets each location pull in
    relevant context from the other modality individually."""
    def __init__(self, in_channels, reduction=8):
        super().__init__()
        self.v_attends_t = CrossModalAttention(in_channels, reduction)
        self.t_attends_v = CrossModalAttention(in_channels, reduction)
        self.conv_out = nn.Conv2d(in_channels * 2, in_channels, 1)

    def forward(self, en_ir, en_vi):
        vi_refined = self.v_attends_t(query_feat=en_vi, kv_feat=en_ir)
        ir_refined = self.t_attends_v(query_feat=en_ir, kv_feat=en_vi)
        fused = torch.cat((ir_refined, vi_refined), dim=1)
        return self.conv_out(fused)


# Fusion strategy, two type
class Fusion_strategy(nn.Module):
    def __init__(self, in_channels):
        super(Fusion_strategy, self).__init__()
        self.fusion_add = Fusion_ADD()
        self.fusion_avg = Fusion_AVG()
        self.fusion_max = Fusion_MAX()
        self.fusion_cat = Fusion_CAT(in_channels=in_channels)
        self.fusion_flf = Fusion_FLF(in_channels=in_channels)
        self.fusion_cmattn = Fusion_CMATTN(in_channels=in_channels)

    def forward(self, v_feat, t_feat, fs_type):
        self.fs_type = fs_type
        if self.fs_type == 'add':
            fusion_operation = self.fusion_add
        elif self.fs_type == 'avg':
            fusion_operation = self.fusion_avg
        elif self.fs_type == 'max':
            fusion_operation = self.fusion_max
        elif self.fs_type == 'cat':
            fusion_operation = self.fusion_cat
        elif self.fs_type == 'flf':
            fusion_operation = self.fusion_flf
        elif self.fs_type == 'cmattn':
            fusion_operation = self.fusion_cmattn
        else:
            raise ValueError(f"Unknown fs_type: {fs_type}")

        if isinstance(v_feat, tuple) or isinstance(v_feat, list):
            fused_feat = []
            for i in range(len(v_feat)):
                fused_feat.append(fusion_operation(v_feat[i], t_feat[i]))
        else:
            fused_feat = fusion_operation(v_feat, t_feat)

        return fused_feat



class PoolingUpsample(nn.Module):
    def __init__(self, in_channels):
        super().__init__()
        self.in_channels = in_channels
        self.maxpooling = nn.MaxPool2d(2, 2, dilation=1)
        # self.upsample = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        self.conv1x1 = nn.Conv2d(in_channels*2, in_channels, 1)

    def forward(self, x):
        x_ = self.maxpooling(x)
        # x_ = self.upsample(x_)
        x_ = F.interpolate(x_, mode='bilinear', size=x.shape[-2:], align_corners=True)
        # import pdb; pdb.set_trace()
        x = self.conv1x1(torch.cat((x, x_), 1))
        return x
```

```python
%%writefile /content/drive/MyDrive/hackathon_project/baseline/mmdet-rgbtdroneperson/mmdet/models/detectors/quality_adaptive_fusion.py

import torch
import torch.nn as nn


class QualityAdaptiveFusion(nn.Module):

    def _init_(self, in_channels):
        super(QualityAdaptiveFusion, self)._init_()

        self.fusion = nn.Sequential(
            nn.Conv2d(in_channels * 2, in_channels, kernel_size=1),
            nn.BatchNorm2d(in_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, rgb_feat, thermal_feat, rgb_quality, thermal_quality):

        # Normalize quality scores
        rgb_quality = torch.sigmoid(rgb_quality)
        thermal_quality = torch.sigmoid(thermal_quality)

        # Calculate total quality
        total_quality = rgb_quality + thermal_quality + 1e-6

        # Calculate adaptive weights
        rgb_weight = rgb_quality / total_quality
        thermal_weight = thermal_quality / total_quality

        # Reweight features
        rgb_feat = rgb_feat * rgb_weight
        thermal_feat = thermal_feat * thermal_weight

        # Combine RGB and thermal features
        fused_feat = torch.cat((rgb_feat, thermal_feat), dim=1)

        # Reduce channels back to original size
        fused_feat = self.fusion(fused_feat)

        return fused_feat
```

```python
%%writefile /content/drive/MyDrive/hackathon_project/baseline/mmdet-rgbtdroneperson/mmdet/models/detectors/qfdet.py
from ..builder import DETECTORS, build_backbone, build_head
from .single_stage import SingleStageDetector
import torch
import torch.nn as nn
import torch.nn.functional as F
from mmdet.core import bbox2result, multi_apply


@DETECTORS.register_module()
class QFDet(SingleStageDetector):
    """Implementation of `QFDet`."""

    def __init__(self,
                 backbone,
                 neck,
                 bbox_head,
                 bbox_prehead,
                 base_fusion='cat',
                 quality_attention=True,
                 poolupsample=None,
                 reweight=False,
                 train_cfg=None,
                 test_cfg=None,
                 pretrained=None,
                 init_cfg=None):
        super(QFDet, self).__init__(backbone, neck, bbox_head, train_cfg,
                                   test_cfg, pretrained, init_cfg)
        self.backbone_t = build_backbone(backbone)
        self.base_fusion = base_fusion
        self.fuse = Fusion_strategy(neck['out_channels'])
        bbox_prehead.update(train_cfg=train_cfg)
        bbox_prehead.update(test_cfg=test_cfg)
        self.bbox_prehead = build_head(bbox_prehead)
        self.iter = 0
        if poolupsample is not None:
            self.poolupsample = PoolingUpsample(neck['out_channels'])
        else:
            self.poolupsample = None
        self.quality_attention = quality_attention

        self.reweight = reweight

    def extract_feat(self, img):
        """Directly extract features from the backbone+neck."""
        # self.iter = self.iter + 1
        v_img, t_img = img
        v_feats = self.backbone(v_img)
        t_feats = self.backbone_t(t_img)
        if self.with_neck:
            v_feats = self.neck(v_feats)
            t_feats = self.neck(t_feats)

        return (v_feats, t_feats)

    def forward_dummy(self, img):
        """Used for computing network flops.

        See `mmdetection/tools/analysis_tools/get_flops.py`
        """
        x = self.extract_feat(img)
        x_vs, x_ts = x
        num_level = len(x_vs)
        quality_preds_t, quality_preds_v = self.bbox_prehead.forward_test(x)
        fused_x = []
        for i in range(num_level):
            quality_pred_t, quality_pred_v = my_norm(quality_preds_t[i], quality_preds_v[i], type='minmax')

            x_t = (1 + quality_pred_t) * x_ts[i]
            x_v = (1 + quality_pred_v) * x_vs[i]

            if self.poolupsample is not None and i < num_level-1:
                x_v = self.poolupsample(x_v)

            fused_x_ = self.fuse(x_t, x_v, 'cat')
            fused_x.append(fused_x_)
        outs = self.bbox_head(fused_x)
        return outs

    def forward_train(self,
                      img,
                      img_metas,
                      gt_bboxes,
                      gt_labels,
                      gt_bboxes_ignore=None):
        """
        Args:
            img (Tensor): Input images of shape (N, C, H, W).
                Typically these should be mean centered and std scaled.
            img_metas (list[dict]): A List of image info dict where each dict
                has: 'img_shape', 'scale_factor', 'flip', and may also contain
                'filename', 'ori_shape', 'pad_shape', and 'img_norm_cfg'.
                For details on the values of these keys see
                :class:`mmdet.datasets.pipelines.Collect`.
            gt_bboxes (list[Tensor]): Each item are the truth boxes for each
                image in [tl_x, tl_y, br_x, br_y] format.
            gt_labels (list[Tensor]): Class indices corresponding to each box
            gt_bboxes_ignore (None | list[Tensor]): Specify which bounding
                boxes can be ignored when computing the loss.

        Returns:
            dict[str, Tensor]: A dictionary of loss components.
        """
        # super(SingleStageDetector, self).forward_train(img, img_metas)
        losses = dict()
        x = self.extract_feat(img)
        x_vs, x_ts = x

        cls_scores_t, bbox_preds_t, centernesses_t, quality_preds_t, \
        cls_scores_v, bbox_preds_v, centernesses_v, quality_preds_v = self.bbox_prehead(x)

        pre_loss = self.bbox_prehead.loss(
             cls_scores_t,
             bbox_preds_t,
             centernesses_t,
             quality_preds_t,
             cls_scores_v,
             bbox_preds_v,
             centernesses_v,
             quality_preds_v,
             gt_bboxes,
             gt_labels,
             img_metas,
             gt_bboxes_ignore=None)
        losses.update(pre_loss)

        fused_x = self.qce_fusion(x, quality_preds_t, quality_preds_t)

        loss = self.bbox_head.forward_train(fused_x, img_metas, gt_bboxes,
                                              gt_labels, gt_bboxes_ignore)
        if self.reweight:
            weight = torch.linspace(1, 5, steps=len(x_ts), device=loss['loss_cls'][0].device)
            for k in loss.keys():
                for i in range(weight.size()[0]):
                    loss[k][i] = weight[i]*loss[k][i]
        losses.update(loss)
        return losses

    def simple_test(self, img, img_metas, rescale=False):
        """Test function without test-time augmentation.

        Args:
            img (torch.Tensor): Images with shape (N, C, H, W).
            img_metas (list[dict]): List of image information.
            rescale (bool, optional): Whether to rescale the results.
                Defaults to False.

        Returns:
            list[list[np.ndarray]]: BBox results of each image and classes.
                The outer list corresponds to each image. The inner list
                corresponds to each class.
        """
        self.iter += 1
        x = self.extract_feat(img)

        quality_preds_t, quality_preds_v = self.bbox_prehead.forward_test(x)

        fused_x = self.qce_fusion(x, quality_preds_t, quality_preds_v)

        results_list = self.bbox_head.simple_test(
            fused_x, img_metas, rescale=rescale)

        bbox_results = [
            bbox2result(det_bboxes, det_labels, self.bbox_head.num_classes)
            for det_bboxes, det_labels in results_list
        ]
        return bbox_results

    def qce_fusion(self, x, quality_t, quality_v):
        x_vs, x_ts = x

        num_level = len(x_vs)

        fused_x = []
        for i in range(num_level):
            x_t = x_ts[i]
            x_v = x_vs[i]
            if self.quality_attention:
                quality_pred_t = torch.max(quality_t[i], dim=1, keepdim=True)[0]
                quality_pred_v = torch.max(quality_v[i], dim=1, keepdim=True)[0]

                quality_pred_t, quality_pred_v = my_norm(quality_pred_t, quality_pred_v, type='minmax')

                x_t = (1 + quality_pred_t) * x_t
                x_v = (1 + quality_pred_v) * x_v

            if self.poolupsample is not None and i < num_level-1:
                x_v = self.poolupsample(x_v)
                # x_t = self.poolupsample(x_t)

            fused_x_ = self.fuse(x_t, x_v, self.base_fusion)

            fused_x.append(fused_x_)

        return fused_x


def save_feature_to_img(features, name, timestamp, method='cv2', channel=None, output_dir=None, maxmin=None):
    import numpy as np
    import cv2
    import matplotlib.pyplot as plt
    import os

    if output_dir is None:
        output_dir = '/home/zhangyan22/mmdetection/work_dirs/atss_r50_fpn_hf_qprelchead_1x_dp/vis_wo_optimize_v'
    if not os.path.exists(output_dir):
        os.mkdir(output_dir)
    if not isinstance(timestamp, str):
        timestamp = str(timestamp)

    # for i in range(len(features)):
    if isinstance(features, list) or isinstance(features, tuple):
        for i in range(3):
            features_ = features[i]
            for j in range(features_.shape[0]):
                upsample = nn.Upsample(scale_factor=4, mode='bilinear', align_corners=True)
                features_ = upsample(features_)

                feature = features_[j, :, :, :]
                if channel is None:
                    feature = torch.sum(feature, 0)
                else:
                    feature = feature[channel, :, :]
                feature = feature.detach().cpu().numpy() # 转为numpy

                dist_dir = os.path.join(output_dir, timestamp)
                if not os.path.exists(dist_dir):
                    os.mkdir(dist_dir)

                if method == 'cv2':
                    if maxmin is not None:
                        img = (feature - maxmin[1])/(maxmin[0] - maxmin[1] + 1e-5) * 255
                    else:
                        img = (feature - np.amin(feature))/(np.amax(feature) - np.amin(feature) + 1e-5) * 255 # 注意要防止分母为0！
                    img = img.astype(np.uint8)
                    img = cv2.applyColorMap(img, cv2.COLORMAP_JET)
                    plt.imshow(feature)
                    plt.axis('off')
                    cv2.imwrite(os.path.join(dist_dir, name + str(i) + '.jpg'), img)

                elif method == 'matshow':
                    plt.matshow(feature, interpolation='nearest')
                    plt.colorbar()
                    plt.axis('off')

                    plt.savefig(os.path.join(dist_dir, name + str(i) + '.png'))
                    plt.close()
                else:
                    NotImplementedError()

    else:
        for j in range(features.shape[0]):
            upsample = nn.Upsample(scale_factor=4, mode='bilinear', align_corners=True)
            features = upsample(features)

            feature = features[j, :, :, :]
            if channel is None:
                feature = torch.sum(feature, 0)
            else:
                feature = feature[channel, :, :]
            feature = feature.detach().cpu().numpy() # 转为numpy

            dist_dir = os.path.join(output_dir, timestamp)
            if not os.path.exists(dist_dir):
                os.mkdir(dist_dir)

            if method == 'cv2':
                if maxmin is not None:
                    img = (feature - maxmin[1])/(maxmin[0] - maxmin[1] + 1e-5) * 255
                else:
                    img = (feature - np.amin(feature))/(np.amax(feature) - np.amin(feature) + 1e-5) * 255 # 注意要防止分母为0！
                img = img.astype(np.uint8)
                img = cv2.applyColorMap(img, cv2.COLORMAP_JET)
                plt.imshow(feature)
                plt.axis('off')
                cv2.imwrite(os.path.join(dist_dir, name + '.jpg'), img)

            elif method == 'matshow':
                plt.matshow(feature, interpolation='nearest')
                plt.colorbar()
                plt.axis('off')

                plt.savefig(os.path.join(dist_dir, name + '.png'))
                plt.close()
            else:
                NotImplementedError()


def feature_map_norm(x):
    # input (B, C, H, W)
    bs, c, h, w = x.shape
    x = x.view(bs, -1, h*w)
    x_mean = torch.mean(x, dim=2, keepdim=True)
    x_std = torch.std(x, dim=2, keepdim=True)
    x = (x - x_mean) / x_std
    x = x.view(bs, c, h, w)
    return x

def my_norm(x1, x2, type='standard'):
    assert type in ['standard', 'minmax']
    bs, _ , H, W = x1.size()
    _, _, h, w = x2.size()
    x1 = x1.view(bs, -1, H*W)
    x2 = x2.view(bs, -1, h*w)
    concat = torch.cat((x1, x2), dim=2)
    if type == 'standard':
        x_mean = torch.mean(concat, dim=2, keepdim=True)
        x_std = torch.std(concat, dim=2, keepdim=True)
        x1 = (x1 - x_mean) / x_std
        x2 = (x2 - x_mean) / x_std
    elif type == 'minmax':
        x_min = torch.min(concat, dim=2, keepdim=True)[0]
        x_max = torch.max(concat, dim=2, keepdim=True)[0]
        x1 = (x1 - x_min) / x_max
        x2 = (x2 - x_min) / x_max
    x1 = x1.view(bs, -1, H, W)
    x2 = x2.view(bs, -1, h, w)
    return [x1, x2]


class Fusion_ADD(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = en_ir + en_vi
        return temp

class Fusion_AVG(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = (en_ir + en_vi) / 2
        return temp

class Fusion_MAX(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = torch.max(en_ir, en_vi)
        return temp

class Fusion_CAT(torch.nn.Module):
    def __init__(self, in_channels) -> None:
        super().__init__()
        self.conv1x1 = nn.Conv2d(2*in_channels, in_channels, 1)

    def forward(self, en_ir, en_vi):
        temp = torch.cat((en_ir, en_vi), 1)
        temp = self.conv1x1(temp)
        return temp

# Fusion strategy, two type
class Fusion_strategy(nn.Module):
    def __init__(self, in_channels):
        super(Fusion_strategy, self).__init__()
        self.fusion_add = Fusion_ADD()
        self.fusion_avg = Fusion_AVG()
        self.fusion_max = Fusion_MAX()
        self.fusion_cat = Fusion_CAT(in_channels=in_channels)

    def forward(self, v_feat, t_feat, fs_type):
        self.fs_type = fs_type
        if self.fs_type == 'add':
            fusion_operation = self.fusion_add
        elif self.fs_type == 'avg':
            fusion_operation = self.fusion_avg
        elif self.fs_type == 'max':
            fusion_operation = self.fusion_max
        elif self.fs_type == 'cat':
            fusion_operation = self.fusion_cat
        if isinstance(v_feat, tuple) or isinstance(v_feat, list):
            fused_feat = []
            for i in range(len(v_feat)):
                fused_feat.append(fusion_operation(v_feat[i], t_feat[i]))
        else:
            fused_feat = fusion_operation(v_feat, t_feat)

        return fused_feat



class PoolingUpsample(nn.Module):
    def __init__(self, in_channels):
        super().__init__()
        self.in_channels = in_channels
        self.maxpooling = nn.MaxPool2d(2, 2, dilation=1)
        # self.upsample = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        self.conv1x1 = nn.Conv2d(in_channels*2, in_channels, 1)

    def forward(self, x):
        x_ = self.maxpooling(x)
        # x_ = self.upsample(x_)
        x_ = F.interpolate(x_, mode='bilinear', size=x.shape[-2:], align_corners=True)
        # import pdb; pdb.set_trace()
        x = self.conv1x1(torch.cat((x, x_), 1))
        return x
```

```python
%%writefile mmdet/models/utils/cbam.py

import torch
import torch.nn as nn

# ------------------------
# Channel Attention
# ------------------------
class ChannelAttention(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()

        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)

        self.mlp = nn.Sequential(
            nn.Conv2d(channels, channels // reduction, 1, bias=False),
            nn.ReLU(inplace=True),
            nn.Conv2d(channels // reduction, channels, 1, bias=False)
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        avg = self.mlp(self.avg_pool(x))
        mx = self.mlp(self.max_pool(x))

        out = avg + mx

        return self.sigmoid(out)


# ------------------------
# Spatial Attention
# ------------------------
class SpatialAttention(nn.Module):

    def __init__(self, kernel_size=7):
        super().__init__()

        padding = kernel_size // 2

        self.conv = nn.Conv2d(
            2,
            1,
            kernel_size,
            padding=padding,
            bias=False
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        avg = torch.mean(x, dim=1, keepdim=True)

        mx, _ = torch.max(x, dim=1, keepdim=True)

        x = torch.cat([avg, mx], dim=1)

        x = self.conv(x)

        return self.sigmoid(x)


# ------------------------
# CBAM Module
# ------------------------
class CBAM(nn.Module):

    def __init__(self, channels):
        super().__init__()

        self.ca = ChannelAttention(channels)
        self.sa = SpatialAttention()

    def forward(self, x):

        x = self.ca(x) * x

        x = self.sa(x) * x

        return x
```

```python
%%writefile mmdet/models/detectors/qfdet.py
# Copyright (c) OpenMMLab. All rights reserved.
from ..builder import DETECTORS, build_backbone, build_head
from .single_stage import SingleStageDetector
import torch
import torch.nn as nn
import torch.nn.functional as F
from mmdet.core import bbox2result, multi_apply
from mmdet.models.utils.cbam import CBAM


@DETECTORS.register_module()
class QFDet(SingleStageDetector):
    """Implementation of `QFDet`."""

    def __init__(self,
                 backbone,
                 neck,
                 bbox_head,
                 bbox_prehead,
                 base_fusion='cat',
                 quality_attention=True,
                 poolupsample=None,
                 reweight=False,
                 train_cfg=None,
                 test_cfg=None,
                 pretrained=None,
                 init_cfg=None):
        super(QFDet, self).__init__(backbone, neck, bbox_head, train_cfg,
                                   test_cfg, pretrained, init_cfg)
        self.backbone_t = build_backbone(backbone)
        self.base_fusion = base_fusion
        self.fuse = Fusion_strategy(neck['out_channels'])

        self.rgb_cbam = CBAM(neck['out_channels'])
        self.thermal_cbam = CBAM(neck['out_channels'])
        bbox_prehead.update(train_cfg=train_cfg)
        bbox_prehead.update(test_cfg=test_cfg)
        self.bbox_prehead = build_head(bbox_prehead)
        self.iter = 0
        if poolupsample is not None:
            self.poolupsample = PoolingUpsample(neck['out_channels'])
        else:
            self.poolupsample = None
        self.quality_attention = quality_attention

        self.reweight = reweight

    def extract_feat(self, img):
        """Directly extract features from the backbone+neck."""
        # self.iter = self.iter + 1
        v_img, t_img = img
        v_feats = self.backbone(v_img)
        t_feats = self.backbone_t(t_img)
        if self.with_neck:
            v_feats = self.neck(v_feats)
            t_feats = self.neck(t_feats)

        return (v_feats, t_feats)

    def forward_dummy(self, img):
        """Used for computing network flops.

        See `mmdetection/tools/analysis_tools/get_flops.py`
        """
        x = self.extract_feat(img)
        x_vs, x_ts = x
        num_level = len(x_vs)
        quality_preds_t, quality_preds_v = self.bbox_prehead.forward_test(x)
        fused_x = []
        for i in range(num_level):
            quality_pred_t, quality_pred_v = my_norm(quality_preds_t[i], quality_preds_v[i], type='minmax')

            x_t = (1 + quality_pred_t) * x_ts[i]
            x_v = (1 + quality_pred_v) * x_vs[i]

            if self.poolupsample is not None and i < num_level-1:
                x_v = self.poolupsample(x_v)

            fused_x_ = self.fuse(x_t, x_v, 'cat')
            fused_x.append(fused_x_)
        outs = self.bbox_head(fused_x)
        return outs

    def forward_train(self,
                      img,
                      img_metas,
                      gt_bboxes,
                      gt_labels,
                      gt_bboxes_ignore=None):
        """
        Args:
            img (Tensor): Input images of shape (N, C, H, W).
                Typically these should be mean centered and std scaled.
            img_metas (list[dict]): A List of image info dict where each dict
                has: 'img_shape', 'scale_factor', 'flip', and may also contain
                'filename', 'ori_shape', 'pad_shape', and 'img_norm_cfg'.
                For details on the values of these keys see
                :class:`mmdet.datasets.pipelines.Collect`.
            gt_bboxes (list[Tensor]): Each item are the truth boxes for each
                image in [tl_x, tl_y, br_x, br_y] format.
            gt_labels (list[Tensor]): Class indices corresponding to each box
            gt_bboxes_ignore (None | list[Tensor]): Specify which bounding
                boxes can be ignored when computing the loss.

        Returns:
            dict[str, Tensor]: A dictionary of loss components.
        """
        # super(SingleStageDetector, self).forward_train(img, img_metas)
        losses = dict()
        x = self.extract_feat(img)
        x_vs, x_ts = x

        cls_scores_t, bbox_preds_t, centernesses_t, quality_preds_t, \
        cls_scores_v, bbox_preds_v, centernesses_v, quality_preds_v = self.bbox_prehead(x)

        pre_loss = self.bbox_prehead.loss(
             cls_scores_t,
             bbox_preds_t,
             centernesses_t,
             quality_preds_t,
             cls_scores_v,
             bbox_preds_v,
             centernesses_v,
             quality_preds_v,
             gt_bboxes,
             gt_labels,
             img_metas,
             gt_bboxes_ignore=None)
        losses.update(pre_loss)

        fused_x = self.qce_fusion(x, quality_preds_t, quality_preds_t)

        loss = self.bbox_head.forward_train(fused_x, img_metas, gt_bboxes,
                                              gt_labels, gt_bboxes_ignore)
        if self.reweight:
            weight = torch.linspace(1, 5, steps=len(x_ts), device=loss['loss_cls'][0].device)
            for k in loss.keys():
                for i in range(weight.size()[0]):
                    loss[k][i] = weight[i]*loss[k][i]
        losses.update(loss)
        return losses

    def simple_test(self, img, img_metas, rescale=False):
        """Test function without test-time augmentation.

        Args:
            img (torch.Tensor): Images with shape (N, C, H, W).
            img_metas (list[dict]): List of image information.
            rescale (bool, optional): Whether to rescale the results.
                Defaults to False.

        Returns:
            list[list[np.ndarray]]: BBox results of each image and classes.
                The outer list corresponds to each image. The inner list
                corresponds to each class.
        """
        self.iter += 1
        x = self.extract_feat(img)

        quality_preds_t, quality_preds_v = self.bbox_prehead.forward_test(x)

        fused_x = self.qce_fusion(x, quality_preds_t, quality_preds_v)

        results_list = self.bbox_head.simple_test(
            fused_x, img_metas, rescale=rescale)

        bbox_results = [
            bbox2result(det_bboxes, det_labels, self.bbox_head.num_classes)
            for det_bboxes, det_labels in results_list
        ]
        return bbox_results

    def qce_fusion(self, x, quality_t, quality_v):
        x_vs, x_ts = x

        num_level = len(x_vs)

        fused_x = []
        for i in range(num_level):
           # Original feature maps
            x_t = x_ts[i]
            x_v = x_vs[i]

# ---------------------------------
# Stage 3: CBAM Enhancement
# ---------------------------------
            x_t = self.thermal_cbam(x_t)
            x_v = self.rgb_cbam(x_v)
            if self.quality_attention:
                quality_pred_t = torch.max(quality_t[i], dim=1, keepdim=True)[0]
                quality_pred_v = torch.max(quality_v[i], dim=1, keepdim=True)[0]

                quality_pred_t, quality_pred_v = my_norm(quality_pred_t, quality_pred_v, type='minmax')

                x_t = (1 + quality_pred_t) * x_t
                x_v = (1 + quality_pred_v) * x_v

            if self.poolupsample is not None and i < num_level-1:
                x_v = self.poolupsample(x_v)
                # x_t = self.poolupsample(x_t)

            fused_x_ = self.fuse(x_t, x_v, self.base_fusion)

            fused_x.append(fused_x_)

        return fused_x


def save_feature_to_img(features, name, timestamp, method='cv2', channel=None, output_dir=None, maxmin=None):
    import numpy as np
    import cv2
    import matplotlib.pyplot as plt
    import os

    if output_dir is None:
        output_dir = '/home/zhangyan22/mmdetection/work_dirs/atss_r50_fpn_hf_qprelchead_1x_dp/vis_wo_optimize_v'
    if not os.path.exists(output_dir):
        os.mkdir(output_dir)
    if not isinstance(timestamp, str):
        timestamp = str(timestamp)

    # for i in range(len(features)):
    if isinstance(features, list) or isinstance(features, tuple):
        for i in range(3):
            features_ = features[i]
            for j in range(features_.shape[0]):
                upsample = nn.Upsample(scale_factor=4, mode='bilinear', align_corners=True)
                features_ = upsample(features_)

                feature = features_[j, :, :, :]
                if channel is None:
                    feature = torch.sum(feature, 0)
                else:
                    feature = feature[channel, :, :]
                feature = feature.detach().cpu().numpy() # 转为numpy

                dist_dir = os.path.join(output_dir, timestamp)
                if not os.path.exists(dist_dir):
                    os.mkdir(dist_dir)

                if method == 'cv2':
                    if maxmin is not None:
                        img = (feature - maxmin[1])/(maxmin[0] - maxmin[1] + 1e-5) * 255
                    else:
                        img = (feature - np.amin(feature))/(np.amax(feature) - np.amin(feature) + 1e-5) * 255 # 注意要防止分母为0！
                    img = img.astype(np.uint8)
                    img = cv2.applyColorMap(img, cv2.COLORMAP_JET)
                    plt.imshow(feature)
                    plt.axis('off')
                    cv2.imwrite(os.path.join(dist_dir, name + str(i) + '.jpg'), img)

                elif method == 'matshow':
                    plt.matshow(feature, interpolation='nearest')
                    plt.colorbar()
                    plt.axis('off')

                    plt.savefig(os.path.join(dist_dir, name + str(i) + '.png'))
                    plt.close()
                else:
                    NotImplementedError()

    else:
        for j in range(features.shape[0]):
            upsample = nn.Upsample(scale_factor=4, mode='bilinear', align_corners=True)
            features = upsample(features)

            feature = features[j, :, :, :]
            if channel is None:
                feature = torch.sum(feature, 0)
            else:
                feature = feature[channel, :, :]
            feature = feature.detach().cpu().numpy() # 转为numpy

            dist_dir = os.path.join(output_dir, timestamp)
            if not os.path.exists(dist_dir):
                os.mkdir(dist_dir)

            if method == 'cv2':
                if maxmin is not None:
                    img = (feature - maxmin[1])/(maxmin[0] - maxmin[1] + 1e-5) * 255
                else:
                    img = (feature - np.amin(feature))/(np.amax(feature) - np.amin(feature) + 1e-5) * 255 # 注意要防止分母为0！
                img = img.astype(np.uint8)
                img = cv2.applyColorMap(img, cv2.COLORMAP_JET)
                plt.imshow(feature)
                plt.axis('off')
                cv2.imwrite(os.path.join(dist_dir, name + '.jpg'), img)

            elif method == 'matshow':
                plt.matshow(feature, interpolation='nearest')
                plt.colorbar()
                plt.axis('off')

                plt.savefig(os.path.join(dist_dir, name + '.png'))
                plt.close()
            else:
                NotImplementedError()


def feature_map_norm(x):
    # input (B, C, H, W)
    bs, c, h, w = x.shape
    x = x.view(bs, -1, h*w)
    x_mean = torch.mean(x, dim=2, keepdim=True)
    x_std = torch.std(x, dim=2, keepdim=True)
    x = (x - x_mean) / x_std
    x = x.view(bs, c, h, w)
    return x

def my_norm(x1, x2, type='standard'):
    assert type in ['standard', 'minmax']
    bs, _ , H, W = x1.size()
    _, _, h, w = x2.size()
    x1 = x1.view(bs, -1, H*W)
    x2 = x2.view(bs, -1, h*w)
    concat = torch.cat((x1, x2), dim=2)
    if type == 'standard':
        x_mean = torch.mean(concat, dim=2, keepdim=True)
        x_std = torch.std(concat, dim=2, keepdim=True)
        x1 = (x1 - x_mean) / x_std
        x2 = (x2 - x_mean) / x_std
    elif type == 'minmax':
        x_min = torch.min(concat, dim=2, keepdim=True)[0]
        x_max = torch.max(concat, dim=2, keepdim=True)[0]
        x1 = (x1 - x_min) / x_max
        x2 = (x2 - x_min) / x_max
    x1 = x1.view(bs, -1, H, W)
    x2 = x2.view(bs, -1, h, w)
    return [x1, x2]


class Fusion_ADD(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = en_ir + en_vi
        return temp

class Fusion_AVG(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = (en_ir + en_vi) / 2
        return temp

class Fusion_MAX(torch.nn.Module):
    def forward(self, en_ir, en_vi):
        temp = torch.max(en_ir, en_vi)
        return temp

class Fusion_CAT(torch.nn.Module):
    def __init__(self, in_channels) -> None:
        super().__init__()
        self.conv1x1 = nn.Conv2d(2*in_channels, in_channels, 1)

    def forward(self, en_ir, en_vi):
        temp = torch.cat((en_ir, en_vi), 1)
        temp = self.conv1x1(temp)
        return temp

# Fusion strategy, two type
class Fusion_strategy(nn.Module):
    def __init__(self, in_channels):
        super(Fusion_strategy, self).__init__()
        self.fusion_add = Fusion_ADD()
        self.fusion_avg = Fusion_AVG()
        self.fusion_max = Fusion_MAX()
        self.fusion_cat = Fusion_CAT(in_channels=in_channels)

    def forward(self, v_feat, t_feat, fs_type):
        self.fs_type = fs_type
        if self.fs_type == 'add':
            fusion_operation = self.fusion_add
        elif self.fs_type == 'avg':
            fusion_operation = self.fusion_avg
        elif self.fs_type == 'max':
            fusion_operation = self.fusion_max
        elif self.fs_type == 'cat':
            fusion_operation = self.fusion_cat
        if isinstance(v_feat, tuple) or isinstance(v_feat, list):
            fused_feat = []
            for i in range(len(v_feat)):
                fused_feat.append(fusion_operation(v_feat[i], t_feat[i]))
        else:
            fused_feat = fusion_operation(v_feat, t_feat)

        return fused_feat



class PoolingUpsample(nn.Module):
    def __init__(self, in_channels):
        super().__init__()
        self.in_channels = in_channels
        self.maxpooling = nn.MaxPool2d(2, 2, dilation=1)
        # self.upsample = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        self.conv1x1 = nn.Conv2d(in_channels*2, in_channels, 1)

    def forward(self, x):
        x_ = self.maxpooling(x)
        # x_ = self.upsample(x_)
        x_ = F.interpolate(x_, mode='bilinear', size=x.shape[-2:], align_corners=True)
        # import pdb; pdb.set_trace()
        x = self.conv1x1(torch.cat((x, x_), 1))
        return x
```
